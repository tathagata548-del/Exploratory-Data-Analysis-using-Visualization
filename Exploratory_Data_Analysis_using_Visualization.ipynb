#**Exploratory Data Analysis using Visualization**


Basic Code to Load Datase :

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("Titanic_Dataset.csv")

# Basic exploration
print(df.info())
print(df.describe())

# Handle missing values
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)


Q1. Create a countplot to show the number of passengers by gender.

Ans: sns.countplot(x='Sex', data=df)
plt.title("Number of Passengers by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()

Q2. Create a bar chart to show the number of passengers in each passenger class using different colors.

Ans: sns.countplot(x='Pclass', data=df, palette='Set2')
plt.title("Passengers in Each Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.show()

Q3. Create a histogram to show the distribution of passenger age.

plt.hist(df['Age'], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

Q4. Create a boxplot to detect outliers in the fare column.

Ans: sns.boxplot(y='Fare', data=df)
plt.title("Boxplot of Fare")
plt.ylabel("Fare")
plt.show()

Q5. Create a scatterplot to a. show the relationship between age and fare. b. analyze the relationship between family size (SibSp + Parch) and fare.

Ans: sns.scatterplot(x='Age', y='Fare', data=df)
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()

Q6. Create a countplot to show survival count by gender.
Ans: sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Survival Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()

Q7. Create a bar chart to show survival by passenger class.

Ans: sns.barplot(x='Pclass', y='Survived', data=df)
plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.show()

Q8. Create a heatmap to show correlation between numerical variables.

Ans: plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()


Q9. Identify which passenger class paid the highest fare on average.

Ans: avg_fare = df.groupby('Pclass')['Fare'].mean()
print(avg_fare)
